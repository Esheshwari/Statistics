# **Distribution of Placement Packages of Final-Year Undergraduate Students**


## **Background**

Campus placements are an important indicator of students' career outcomes and institutional performance. Placement packages vary depending on academic performance, skills, internships, and the recruiting company. Analyzing the distribution of placement packages helps understand the salary trends among graduating students.

## **Research Problem**

How are placement packages distributed among final-year undergraduate students?

## **Research Objectives**

- To study the distribution of placement packages offered to final-year students.
- To construct a grouped frequency distribution.
- To calculate relative and cumulative frequencies.
- To visualize the distribution using a histogram.
- To interpret the overall placement package trend.



## **Variable**

Placement Package (Annual CTC in LPA)

Type:

- Quantitative
- Continuous

Unit:

Lakhs Per Annum (LPA)

## **Sample**

100 placed final-year students selected from different departments.

```python
import pandas as pd

df = pd.read_csv(
    "placement_packages.csv",
    header=None,
    names=["Package"]
)

df.head()
```

## **Create Class Intervals**

```python
bins = np.arange(2, 38, 2)

bins
```

## **Frequency Distribution**

```python
frequency = pd.cut(
    df["Package"],
    bins=bins,
    right=False
).value_counts().sort_index()

frequency
```

## **Relative Frequency**

```python
relative_frequency = frequency / frequency.sum()
```

## **Cumulative Frequency**

```python
cumulative_frequency = frequency.cumsum()
```

## **Final Frequency Table**

```python
freq_table = pd.DataFrame({
    "Class Interval": frequency.index.astype(str),
    "Frequency": frequency.values,
    "Relative Frequency": relative_frequency.values,
    "Cumulative Frequency": cumulative_frequency.values
})

freq_table
```

## **Histogram**

```python
plt.figure(figsize=(10,6))

plt.hist(
    df["Package"],
    bins=bins,
    edgecolor="black"
)

plt.title("Distribution of Placement Packages")
plt.xlabel("Placement Package (LPA)")
plt.ylabel("Number of Students")

plt.grid(alpha=0.3)

plt.show()
```

## **Frequency Polygon**

```python
midpoints = (bins[:-1] + bins[1:]) / 2

plt.figure(figsize=(10,6))

plt.plot(
    midpoints,
    frequency.values,
    marker='o'
)

plt.title("Frequency Polygon")
plt.xlabel("Placement Package (LPA)")
plt.ylabel("Frequency")

plt.grid(alpha=0.3)

plt.show()
```

## **Ogive (Cumulative Frequency Curve)**

```python
upper_limits = bins[1:]

plt.figure(figsize=(10,6))

plt.plot(
    upper_limits,
    cumulative_frequency,
    marker='o'
)

plt.title("Ogive")
plt.xlabel("Upper Class Boundary (LPA)")
plt.ylabel("Cumulative Frequency")

plt.grid(alpha=0.3)

plt.show()
```

## **Mean**

```python
mean = df["Package"].mean()
print(f"Mean: {mean:.2f} LPA")
```

## **Median**

```python
median = df["Package"].median()
print(f"Median: {median:.2f} LPA")
```

## **Mode**

```python
mode = df["Package"].mode()

print("Mode:")
print(mode)
```

## **Variance**

```python
variance = df["Package"].var()

print(f"Variance: {variance:.2f}")
```

## **Standard Deviation**

```python
std = df["Package"].std()

print(f"Standard Deviation: {std:.2f}")
```

## **Summary**

```python
print("="*45)

print(f"Mean                : {mean:.2f} LPA")
print(f"Median              : {median:.2f} LPA")
print(f"Mode                : {mode.iloc[0]:.2f} LPA")
print(f"Variance            : {variance:.2f}")
print(f"Standard Deviation  : {std:.2f}")

print("="*45)
```

## **Conclusion**

The analysis indicates that placement packages are concentrated in the mid-range salary brackets, with relatively few high-value offers. The grouped frequency distribution, histogram, and descriptive statistics collectively show a right-skewed distribution, which is commonly observed in campus placement data. This suggests that the majority of students receive competitive entry-level salaries, while a smaller number secure premium packages from top recruiters.
